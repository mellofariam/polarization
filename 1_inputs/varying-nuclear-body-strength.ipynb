```python
import pandas
import numpy as np
```

```python
force_field = pandas.read_csv(
    "ff_nucleus-compartment-level-undivided.csv"
)
force_field.index = force_field.columns
```

```python
force_field.loc[:, ["SP", "NC", "LM"]]
```

```python
force_field.loc[["SP", "NC", "LM"], :]
```

```python
for divide_by in [3, 5, 8]:
    new_force_field = force_field.copy()
    new_force_field.loc[:, ["SP", "NC", "LM"]] /= divide_by
    new_force_field.loc[["SP", "NC", "LM"], :] /= divide_by

    new_force_field.to_csv(
        f"ff_nucleus-compartment-level_divided-by-{divide_by}.csv",
        index=False,
        float_format="%+.6f",
    )
```
