```python
import pandas
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import seaborn as sns
import h5py

import chroma

plt.style.use("chroma.tex")
```

# nuclear bodies as fields

## chr1


```python
chromosome = 1
folder_num = 4 if chromosome == 1 else 5
replica = 1
num_replicas = 32

data_path = f"/scratch/mm146/Polarization/{folder_num}_IMR-90_hg38_chr{chromosome}/4_analysis/lamina-contact/data/complete/lamina-contact-replica{replica}.h5"
```

```python
with h5py.File(data_path, "r") as f:
    lamina_contact = f["lamina-contact"][:]
    B_indices = f["B-indices"][:]
```

```python
fig, ax = plt.subplots(figsize=(2.3, 1.8))

ax.imshow(lamina_contact, aspect="auto", cmap="Reds")

ax.set_xlabel("Bead index")
ax.set_ylabel("Frames")
```

```python
fig, ax = plt.subplots(figsize=(2.3, 1.8))

ax.hist(lamina_contact.mean(axis=0), bins=50, color="silver", edgecolor="black", density=True)

ax.set_xlabel("Fraction in contact with lamina")
ax.set_ylabel("Probability density")
```

```python
data_path = f"/scratch/mm146/Polarization/{folder_num}_IMR-90_hg38_chr{chromosome}/4_analysis/lamina-contact/data/complete/"

fraction_in_contact = []
for replica in range(1, num_replicas + 1):
    with h5py.File(data_path + f"lamina-contact-replica{replica}.h5", "r") as f:
        lamina_contact = f["lamina-contact"][:]
        fraction_in_contact.extend(lamina_contact.mean(axis=0))
```

```python
fig, ax = plt.subplots(figsize=(2.3, 1.8))

ax.hist(
    fraction_in_contact,
    bins=50,
    color="silver",
    edgecolor="black",
    density=True,
)

ax.set_xlabel("Fraction in contact with lamina")
ax.set_ylabel("Probability density")
```
